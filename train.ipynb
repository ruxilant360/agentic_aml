```python
import pandas as pd
from sklearn.model_selection import RandomizedSearchCV

from src.training.train import train_model
from src.training.model import define_model
from src.training.preprocessing import encode_categorical_features, split_train_test
from scipy.stats import randint, uniform
```

```python
df = pd.read_csv('notebooks/data.csv')
```

```python
df = encode_categorical_features(df)
```

```python
X_train, X_test, y_train, y_test = split_train_test(df)
```

```python
def scale_weights(y):
  neg = (y == 0).sum()
  pos = (y == 1).sum()

  weight = neg / pos
  return weight

weight = scale_weights(y_train)

my_params = { "objective" : 'binary:logistic',
        "eval_metric" : 'aucpr',
        "scale_pos_weight" : weight,
        "tree_method" : 'hist',
        "enable_categorical" : True
          }

param_dist = {
    'n_estimators' : randint(100, 500),
     'max_depth': randint(3, 10),
     'min_child_weight' : randint(1, 20),
      'learning_rate': uniform(0.01, 0.3),
      'subsample': uniform(0.6 , 0.4) }
```

```python
result = train_model(X_train=X_train, y_train=y_train, param_distributions=param_dist, base_params=my_params)
```
